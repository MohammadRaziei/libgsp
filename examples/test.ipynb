```python
html_path = 'hybrid-example.html'
with open(html_path) as f:
    html_string = f.read()
```

```python
from IPython.display import display, HTML, IFrame
import base64

def display_html(html_string, width=800, height=500):
    # Encode to base64 for data URL
    html_base64 = base64.b64encode(html_string.encode('utf-8')).decode('utf-8')
    data_url = f"data:text/html;base64,{html_base64}"
    display(IFrame(src=data_url, width=800, height=500))
#     display(HTML(f'''<iframe src="{data_url}" width="{width}" height="{height}" frameborder="0"></iframe>'''))

display_html(html_string)
```
